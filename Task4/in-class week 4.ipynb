```python
# q 4.4

import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

gauss = np.random.normal(185, 8, 100000)
plt.hist(gauss, bins=40, label=f"mean: 185, std: 8")
# vertical line which corresponds to the sample mean:
plt.axvline(x=185,color='red', linestyle='--') # Here 'y' goes by default from 0(bottom) to 1 (top) of the graph.
# shading area representing plus-minus one standard deviation wrt the mean:
plt.axvspan(185 - 8, 185 + 8, color='yellow', alpha=0.3) # alpha = opacidad del sombreado.

plt.legend()
plt.show()


# Calculating the mean and standard deviation:
mean = np.mean(gauss)
print(f"Mean: {mean:.2f}")
std = np.std(gauss)
print(f"Standard Deviation: {std:.2f}")

# CUMULATIVE DISTRIBUTION FUNCTION -> area to the LEFT under the curve 

# probability of any values that is smaller than and equal to the height
cdf_value = norm.cdf(160, loc=mean, scale=std)
print(f"Probability of being shorter or equal than me: {cdf_value:.4f}")

# probability of a height located within 1 standard deviation range

upper = mean + std
lower = mean - std
cdf_upper = norm.cdf(upper, loc=mean, scale=std)
cdf_lower = norm.cdf(lower, loc=mean, scale=std)
prob = cdf_upper - cdf_lower
print(f"Probability of a height located within 1 standard deviation range: {prob:.4f}")
```

```python
# q 4.12
import math
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,3,3,1,1])
y = np.array([1,1,3,3,1])

plt.plot(x, y, label="Original square") # Non-rotated square

# Let's rotate the square 30deg

angle_deg = -30          # Negative means clockwise rotation
theta = math.radians(angle_deg)

rotor = math.cos(theta) + 1j * math.sin(theta)      # Complex rotor
z = x + y*1j             # We write each point/position (x,y) as x + jy. IMPORTANT: just y*j doesn't work!
z_rot = rotor * z        # Multiply vector (z) by the complex rotor

# The result is a complex number, we take real part for the rotated x and the imaginary part for the rotated y
x_rot = np.real(z_rot)
y_rot = np.imag(z_rot)

plt.plot(x_rot , y_rot, color="cyan", label="rotated square")

plt.gca().set_aspect('equal', adjustable='box')          # gca -> "Get Current Axes", .set_aspect(): modifies scailing of subplot, getting axes to be equal means that 1 unit-length in x-axis = 1 unit-length in y-axis. "box" -> leaves white space around figure 
plt.title('Rotation using Complex Rotor')

plt.legend()
plt.show()
```

```python
def scaling(value, bscale=1, bzero=0):
    """ Scale a value with bscale and add offset bzero"""
    result1 = bscale*value + bzero
    result2 = bscale * value
    return result1, result2

val = 3
r1, r2 = scaling(val, 2, 0.01)
print(r1, r2)
```

```python
# q 4.14

import numpy as np

def funcfilter(func, seq):     # func: a function, seq: single number / sequence
    array = np.asarray(seq)
    fil_ter = array[array > 0]
    result = func(fil_ter)

    return result

sq = np.array([-100, -10, 0, 10, 100])
result = funcfilter(np.log10, sq)
print("Result log10:", result)

single_number = 100
result2 = funcfilter(np.log10, single_number)
print("Result log10(100):", result2)
```

```python
# q 4.16

import numpy as np
from scipy import integrate
from astropy.cosmology import Planck18

H0 = Planck18.H0.value
Om0 = Planck18.Om0
Ode0 = Planck18.Ode0

_1Mpc_in_km = 3.086e19 
H0_SI = H0 / _1Mpc_in_km

def H(z):
    f = H0_SI * np.sqrt(Om0 * (1+z)**3 + Ode0)
    return f

def tL(z):
    intg = 1 / ((1 + z) * H(z))
    return intg

a = 0
z = 2       # integral upper bound
tL_secs = integrate.quad(tL, a, z)[0]    # quad() returns a tuple with the area and the estimated error in the calculation. By adding [0] it only returns result.

# Convert to Gyr
secs_to_Gyr = 3.156e16
tL_Gyr = tL_secs / secs_to_Gyr

print(f"Lookback time to z={z}: {tL_Gyr:.4f} Gyr")

```

```python
# q 4.21

import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

np.random.seed(0) # Generator

X = np.random.normal(0, 1, 100000)
Y = np.random.normal(2, 1, 100000)

plt.figure(figsize=(10,6))    # Width 10 inches, height 6 inches

# Histograms
plt.hist(X, bins=30, density=True, alpha=0.5, color='blue', edgecolor='black', label='X')   # (alpha goes from 0 to 1)
plt.hist(Y, bins=30, density=True, alpha=0.5, color='red', edgecolor='black', label='Y')

x_vals = np.linspace(-6, 8, 10000)

# PDF's
plt.plot(x_vals, norm.pdf(x_vals, 0, 1), 'k:', linewidth=4, label='X (PDF)') # : (dotted)
plt.plot(x_vals, norm.pdf(x_vals, 2, 1), 'k--', linewidth=4, label='y (PDF)') # -- (dashed)

# Annotations
plt.annotate(r'$\mathcal{N}(0,1)$', color='blue', xy=(-2.2, 0.15), xytext=(-4, 0.15), arrowprops=dict(arrowstyle='->', color='blue', linewidth=3))
plt.annotate(r'$\mathcal{N}(2,1)$', color='red', xy=(4, 0.15), xytext=(4.5, 0.15), arrowprops=dict(arrowstyle='->', color='red', linewidth=3))

plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("1D Histogram + PDF")
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# data from previous histogram
np.random.seed(0)
X = np.random.normal(0, 1, 100000)
Y = np.random.normal(2, 1, 100000)

plt.figure(figsize=(10, 7))

# Create the 2D Histogram
counts, xedges, yedges, image= plt.hist2d(X, Y, bins=30, cmap='Blues') # (x, y): the input values, bins: number of bins in each dimension (30x30=900 bins)
# The histogram returns a tuple of 3 values. Counts=datapoints per pixel/bin, xedges and yedges=coord of each bin, we use this to be able to determine position of contour lines.

plt.colorbar() # add the colorbar

# To calculate the contour Levels (1-sigma, 2-sigma, 3-sigma), we need to find where the 68%, 95%, and 99.7% of the data is.
counts_flattened = np.sort(counts.flatten())[::-1] # Counts is a 2D matrix (x, y) values, with .flatten() what we achieve is to make it a 1D array. We then sort counts in descending order.
total_counts = counts_flattened.sum()
cumulative_sum = np.cumsum(counts_flattened) # The cumulative sum here is used to go from inside to outside adding new values to the adding each new value to the total of the values that came before it.

# Sigma levels
sigma_1 = counts_flattened[np.where(cumulative_sum < 0.68 * total_counts)[0][-1]] # last value that fits inside the percentage
sigma_2 = counts_flattened[np.where(cumulative_sum < 0.95 * total_counts)[0][-1]]
sigma_3 = counts_flattened[np.where(cumulative_sum < 0.997 * total_counts)[0][-1]]


# Plot contour lines: we first have to find the center of the x, y bins -> left edge + right edge / 2
x_bin_centers = (xedges[:-1] + xedges[1:]) / 2
y_bin_centers = (yedges[:-1] + yedges[1:]) / 2

contour = plt.contour(x_bin_centers, y_bin_centers, counts.T, 
                      levels=[sigma_3, sigma_2, sigma_1], 
                      colors='red')    # we use the transpose os the counts matrix because matplotlib function .countour() expects input matrix in this format (Y, X)

fmt = {sigma_1: '1σ', sigma_2: '2σ', sigma_3: '3σ'}
plt.clabel(contour, inline=True, fontsize=15, fmt=fmt)

# Details
plt.title("2D Histogram + Contour")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(color='black', linestyle='--', alpha=0.5)

plt.gca().set_aspect('equal')      
plt.show()
```
