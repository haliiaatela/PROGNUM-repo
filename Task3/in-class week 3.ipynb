```python
# q 3.2

masses = [1.9891e+30, 1.8986e+27, 5.6846e+26, 
          1.0243e+26, 8.6810e+25, 5.9736e+24, 
          4.8685e+24, 6.4185e+23, 3.3022e+23, 
          7.349e+22, 1.25e22]
radii  = [696340, 71492, 60268, 24764, 
          25559, 6378.1, 6051.8, 3396.2, 
          2439.7, 1738.1, 1195]
names  = ['Sun', 'Jupiter', 'Saturn', 'Neptune', 
          'Uranus', 'Earth', 'Venus', 'Mars', 
          'Mercury', 'Moon', 'Pluto']


import math
from scipy.constants import G

def velocityescape (radii, masses):
    v_esc = []
    for i in range(len(masses)):
        k = math.sqrt(2*G*masses[i] / radii[i])
        v_esc.append(k)
    return v_esc

result = velocityescape(radii, masses)
print("Escape velocity =", result)

print()

for i in range(len(masses)):
    s = f"Mass of {names[i]} is: {masses[i]:.2f}"
    print(s) 

print()

dictionary = {names[i]: result[i] for i in range(len(masses))} 
print(dictionary)

   



```

```python
# q 3.5

# Long code
list_ = []

for x in range(1, 10):
    S = 1 / x**2
    list_.append(S)
print(list_)


# Abreviate version

list_ = [1 / x**2 for x in range (1, 10)]
print(list_)


# Use of range to check the convergence of a function

N = 10000
convg = 0

for i in range(1, N + 1):
    convg += 1 / i**2
print(convg)   # As we obtain a finite value we know that the sum this sequence converges! 

# the more abreviate version is:

N = 10000
convg = sum(1 / i**2 for i in range(1, N + 1))
print(convg)
```

```python
# q 3.10

from math import pi

def cylinder(r, h):
    Area = 2 * pi * r**2 + 2 * pi * r * h
    Volume = pi * r**2 * h
    return Area, Volume
r = float(input("radius: " ))
h = float(input("height: "))

Area, Volume = cylinder(r, h)

print (f"Area: {Area}, Volume: {Volume}")

print()

ratio = Area / Volume
print(ratio)
```

```python
# q 3.17

import numpy as np

# input array
my_data = np.array([10, 20, 30, 40, 50, 60])


# defining my function
mean = np.sum(my_data) / len(my_data)
print(mean)

def stdev(my_data):
    standard_dev = np.sqrt(np.sum((my_data - mean)**2) / len(my_data))
    return standard_dev
my_result = stdev(my_data)
print(my_result)

# comparing results
sd = np.std(my_data)
print(sd)

# changing code into a one-liner
a = 1 / len(my_data)
b = np.sum(my_data) 
c = a * b # mean
d = a * np.sum((my_data - c)**2))

sd_1line = np.sqrt(d)
print(sd_1line)
          
          

```

```python
# q 3.18

import numpy as np

x = np.array([18.49, 35.48, 35.48, 11.34, 7.12, 9.82, 23.66, 44, 66.7, 52.4, 55, 25.6, 26.19]) # distances
y = np.array([2094.75, 2320.87, 1224.9, 872.13, 432.2, 751, 1410, 2700, 5292, 2550, 5253, 2419, 2158]) # velocities

x_mean = np.mean(x)
y_mean = np.mean(y)

b = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean)**2)
a = y_mean - b * x_mean

print(f"a = {a:.2f}, b = {b:.2f}") # values of incercept and slope with 2 decimal places

```

```python
# q 3.22

from matplotlib import pyplot as plt
import numpy as np

x_range = list(range(-5,6))
y_range = np.cosh(x_range)

# plotting
plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, 'gx--', label="Catenary Curve")

plt.xlabel("x-axis", fontsize=12)
plt.ylabel("y-axis", fontsize=12)
plt.title("Plotting a catenary", fontsize=15, color='grey')

plt.legend()
plt.grid()
plt.show()

#Numpy
x_numpy = np.arange(-5,7)
y_numpy = np.cosh(x_numpy)

plt.figure(figsize=(10, 6))
plt.plot(x_range, y_range, marker="o", color="cyan", label="Catenary Curve")

plt.xlabel("x-axis", fontsize=12)
plt.ylabel("y-axis", fontsize=12)
plt.title("Plotting a catenary", fontsize=15, color='grey')

plt.legend()
plt.grid()
plt.show()

```
