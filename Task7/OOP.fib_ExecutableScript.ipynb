```python
class Fibonacci:
    """Class for calculating Fibonacci sequence"""
    
    def get_nth_term(self, N):
        x = [0,1]
        while len(x)<N:
            x.append(x[-2]+x[-1])
        return x[-1]

    def divisible_by_M(self, N, M):
        x = [0,1]
        while len(x)<N:
            x.append(x[-2]+x[-1])

        # Numbers that can be divided by M
        result=[]
        for value in x:
            if value%M ==0:
                result.append(value)
        return result

fib = Fibonacci()
```
