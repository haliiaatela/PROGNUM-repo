```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import norm

np.random.seed(0)

X = np.random.normal(0, 1, 100000)
Y = np.random.normal(2, 1, 100000)

plt.figure(figsize=(10, 6))

# 1. Plot Histograms
plt.hist(X, bins=30, density=True, alpha=0.5, color='blue', edgecolor='black', label='X')
plt.hist(Y, bins=30, density=True, alpha=0.5, color='red', edgecolor='black', label='Y')

x_vals = np.linspace(-6, 8, 10000)

# 2. Plot PDFs with specific line styles and thicknesses
plt.plot(x_vals, norm.pdf(x_vals, 0, 1), 'k:', linewidth=4, label='X(PDF)') # Dotted
plt.plot(x_vals, norm.pdf(x_vals, 2, 1), 'k--', linewidth=4, label='Y(PDF)') # Dashed

# 3. Enhanced Annotations with Block Arrows
plt.annotate(r'$\mathcal{N}(0, 1)$', xy=(-0.5, 0.15), xytext=(-5, 0.15),
             fontsize=14, color='blue', va='center',
             arrowprops=dict(facecolor='blue', edgecolor='blue', shrink=0.05, width=5, headwidth=15))

plt.annotate(r'$\mathcal{N}(2, 1)$', xy=(4.5, 0.15), xytext=(7.5, 0.15),
             fontsize=14, color='red', va='center', ha='right',
             arrowprops=dict(facecolor='red', edgecolor='red', shrink=0.05, width=5, headwidth=15))

# 4. Axes, Labels, and Grid
plt.xlabel("Values")
plt.ylabel("Probability Density")
plt.title("1D Histogram + PDF")
plt.xlim(-6, 8)
plt.xticks(np.arange(-6, 9, 1))
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()

plt.show()
```

```python
# 4.21


# Create the 2D Histogram
counts, xedges, yedges, image= plt.hist2d(X, Y, bins=30, cmap='Blues') # (x, y): the input values, bins: number of bins in each dimension (30x30=900 bins)
# The histogram returns a tuple of 3 values. Counts=datapoints per pixel/bin, xedges and yedges=coord of each bin, we use this to be able to determine position of contour lines.

plt.colorbar() # add the colorbar

# To calculate the contour Levels (1-sigma, 2-sigma, 3-sigma), we need to find where the 68%, 95%, and 99.7% of the data is.
counts_flattened = np.sort(counts.flatten())[::-1] # Counts is a 2D matrix (x, y) values, with .flatten() what we achieve is to make it a 1D array. We then sort counts in descending order.
total_counts = counts_flattened.sum()
cumulative_sum = np.cumsum(counts_flattened) # The cumulative sum here is used to go from inside to outside adding new values to the adding each new value to the total of the values that came before it.

# Sigma levels
sigma_1 = counts_flattened[np.where(cumulative_sum < 0.68 * total_counts)[0][-1]] # last value that fits inside the percentage
sigma_2 = counts_flattened[np.where(cumulative_sum < 0.95 * total_counts)[0][-1]]
sigma_3 = counts_flattened[np.where(cumulative_sum < 0.997 * total_counts)[0][-1]]


# Plot contour lines: we first have to find the center of the x, y bins -> left edge + right edge / 2
x_bin_centers = (xedges[:-1] + xedges[1:]) / 2
y_bin_centers = (yedges[:-1] + yedges[1:]) / 2

contour = plt.contour(x_bin_centers, y_bin_centers, counts.T, 
                      levels=[sigma_3, sigma_2, sigma_1], 
                      colors='red')    # we use the transpose os the counts matrix because matplotlib function .countour() expects input matrix in this format (Y, X)

fmt = {sigma_1: '1σ', sigma_2: '2σ', sigma_3: '3σ'}
plt.clabel(contour, inline=True, fontsize=15, fmt=fmt)

# Details
plt.title("2D Histogram + Contour")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(color='black', linestyle='--', alpha=0.5)

plt.show()
```
